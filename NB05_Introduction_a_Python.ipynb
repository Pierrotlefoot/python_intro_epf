## 1. Tester, ce n'est pas seulement exécuter

Un programme peut s'exécuter sans erreur et produire un **résultat faux**. Un test compare une **attente déterminée avant l'exécution** au résultat obtenu.

Voici trois fonctions de référence. Identifiez leurs entrées, leur sortie et leur hypothèse éventuelle sur les données.

### Lire un contrat avant de lire les instructions
Une fonction associe des entrées à un résultat selon une règle. Son **contrat** décrit les données admises, les hypothèses et la valeur renvoyée. Il ne raconte pas encore la boucle qui réalise le calcul.

| Fonction | Entrées et hypothèses | Sortie promise |
|---|---|---|
| `compter_anomalies` | liste de nombres, éventuellement vide ; seuil numérique | entier : nombre de valeurs strictement supérieures au seuil |
| `calculer_moyenne` | liste non vide de nombres | nombre : somme des valeurs divisée par leur nombre |
| `trouver_maximum` | liste non vide de nombres | nombre : plus grande valeur présente dans la liste |

Le paramètre `mesures` est un nom local ; l'argument est la liste donnée lors de l'appel. `return` fournit une valeur au programme appelant et termine la fonction. Aucun affichage n'est nécessaire pour qu'un calcul ait un résultat. Ces fonctions lisent la liste sans la modifier.

**Lecture guidée de la cellule suivante :** repérez l'initialisation, le parcours, la mise à jour et le retour. Pour chacune, formulez ce que représente la variable après les éléments déjà parcourus : `compteur` compte les dépassements observés ; `total` contient leur somme numérique, y compris les valeurs qui ne dépassent aucun seuil ; `maximum` contient le plus grand candidat rencontré. Les trois variables ne remplissent donc pas la même responsabilité.


````python
def compter_anomalies(mesures, seuil):
    compteur = 0
    for mesure in mesures:
        if mesure > seuil:
            compteur += 1
    return compteur

def calculer_moyenne(mesures):
    total = 0
    for mesure in mesures:
        total += mesure
    return total / len(mesures)

def trouver_maximum(mesures):
    maximum = mesures[0]
    for mesure in mesures:
        if mesure > maximum:
            maximum = mesure
    return maximum
````

Le comptage possède un résultat naturel sur une liste vide : aucun événement, donc zéro. Une moyenne sans mesure n'est pas une moyenne de zéro ; un maximum sans valeur n'a pas de candidat. ## 3. Dictionnaires : représenter un objet par ses propriétés

Une liste convient pour une **suite de mesures**. Mais un capteur possède plusieurs informations de nature différente : identifiant, température, seuil, état. Les stocker dans une liste obligerait à se souvenir que « l'indice 2 est le seuil ». Un **dictionnaire** associe chaque **clé** à une **valeur**.

```python
capteur = {"id": "T01", "temperature": 42.5, "seuil": 50, "actif": True}
```

On accède à une propriété par sa **clé**, non par sa position. Le dictionnaire est modifiable.

**Prédisez :** que valent `capteur["temperature"]` et `capteur["actif"]` ? Que se passe-t-il si on demande `capteur["pression"]` ?

### Deux questions, deux structures
Une liste répond à « quelle valeur à cette position ? » ; un dictionnaire répond à « quelle valeur pour cette propriété ? ». Dans une liste, `[0]` désigne le premier élément. Dans notre dictionnaire, `["id"]` désigne la valeur associée à la clé `"id"`. Les clés sont ici des chaînes, les valeurs peuvent être une chaîne, un nombre ou un booléen.

Un dictionnaire ne déduit rien des noms de clés : `"temperature"` et `"Temperature"` sont deux clés différentes. Accéder directement à une clé absente provoque `KeyError`. Dans ce notebook, les fonctions supposent des dictionnaires possédant les clés annoncées et des valeurs de types cohérents ; la validation de données externes serait un autre problème.

### Lire et écrire une propriété
Dans `capteur["temperature"] = 51.0`, la partie droite fournit la nouvelle valeur, et la partie gauche désigne l'emplacement à modifier. Une clé existante est mise à jour ; une clé nouvelle est ajoutée. L'affectation ne crée pas un nouveau capteur et ne change pas automatiquement son seuil. Deux clés identiques ne permettent pas de conserver deux valeurs indépendantes : une clé désigne une seule valeur à un instant donné.

Avant la cellule suivante, notez l'état initial puis l'état après chaque affectation. Les propriétés qui ne sont pas visées conservent leur valeur. Pour examiner la clé absente, écrivez éventuellement l'accès dans une cellule séparée : cette erreur volontaire ne doit pas interrompre votre parcours normal.


````python
capteur = {"id": "T01", "temperature": 42.5, "seuil": 50, "actif": True}
print(capteur["temperature"])
capteur["temperature"] = 51.0
capteur["unite"] = "°C"
print(capteur)
````

### Fonction recevant un dictionnaire

Une fonction peut recevoir un dictionnaire comme n'importe quelle autre valeur. Pour préserver une séparation claire des responsabilités, elle **retourne** un diagnostic au lieu de l'afficher.

**Contrat de `est_en_alerte` :** recevoir un dictionnaire contenant `actif` (booléen), `temperature` et `seuil` (nombres), et renvoyer un booléen. La règle combine deux exigences : le capteur est actif **et** sa température dépasse strictement son seuil. Il faut satisfaire les deux.

Dans la cellule précédente, la température du capteur a été modifiée : l'appel utilise son état courant. Une fonction ne reçoit pas automatiquement une copie du dictionnaire. Celle-ci ne fait que lire ses propriétés ; une affectation à une de ses clés modifierait l'objet partagé avec le programme appelant.

**Lire le `and` :** Python évalue d'abord la condition de gauche. Si elle est fausse, la condition complète est déjà fausse et la partie droite n'est pas évaluée. Ici, un capteur désactivé ne déclenche pas d'alerte ; sa température reste une donnée stockée, mais elle est exclue de cette décision. Une température exactement égale au seuil ne constitue pas un dépassement.


````python
def est_en_alerte(capteur):
    return capteur["actif"] and capteur["temperature"] > capteur["seuil"]

print(est_en_alerte(capteur))
````

## 7. Prolongement facultatif — `while` (si le temps le permet)

Avec `for`, on parcourt une séquence ou un ensemble d'itérations déterminé. Avec `while`, on répète **tant qu'une condition reste vraie**. Le nombre de répétitions peut être inconnu au départ.

Trois questions à poser avant d'écrire une boucle `while` : **quel état initial ? quelle condition ? quelle instruction fait évoluer l'état vers l'arrêt ?**

### La condition autorise le prochain passage
Python évalue la condition **avant** chaque passage. Si elle est vraie, il exécute le corps entier, puis revient au test ; si elle est fausse, il continue après la boucle. Si elle est fausse dès le départ, le corps ne s'exécute jamais. `while` n'arrête pas le corps à l'instant où une variable dépasse un seuil : le nouvel état est examiné au prochain test.

`for` convient lorsque l'on veut visiter les éléments d'une collection. `while` convient lorsque la poursuite dépend de l'état courant. Son écriture demande de justifier l'arrêt : quelle instruction fait évoluer l'état, et pourquoi cette évolution doit-elle rendre la condition fausse ? Ici, l'incrément positif permet d'atteindre un seuil fini.

Avant d'exécuter l'exemple, tracez `charge` avant le test, la valeur de `charge < 80`, puis `charge` et `iterations` après le corps. Le premier passage part de 15 et finit à 35 avec une itération ; poursuivez vous-même. Distinguez l'état final du seuil visé : un pas discret peut dépasser le seuil sans tomber exactement dessus.

**Avant la cellule suivante : prédisez le résultat final et le nombre de passages.**


````python
charge = 15
iterations = 0
while charge < 80:
    charge += 20
    iterations += 1

print(charge, iterations)
````
