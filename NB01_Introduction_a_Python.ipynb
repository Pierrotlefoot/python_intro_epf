## 3. Git et GitHub : conserver les étapes de son travail
**Git** suit les versions d'un projet sur votre ordinateur. Un **dépôt** (*repository*) réunit les fichiers et leur historique. Un **commit** est une étape enregistrée dans cet historique.

**GitHub** héberge des dépôts distants et facilite leur partage. Un dépôt **local** est sur votre machine ; un dépôt **distant** est accessible en ligne. Un commit reste local tant que vous ne l'avez pas envoyé.

Notre circuit : **modifier → enregistrer le fichier → préparer les changements → créer un commit → envoyer sur GitHub**.

### Première utilisation : récupérer son dépôt
L'enseignant vous indique le dépôt à utiliser et vous donne un accès en écriture à votre espace, ou vous fait créer un **fork**, une copie du dépôt sur votre compte GitHub.

Dans les commandes ci-dessous, remplacez `URL_DE_VOTRE_DEPOT` par l'URL réelle de **votre dépôt de travail**. Le nom `cours-python` sera celui du nouveau dossier local. N'exécutez `clone` qu'une fois pour cette copie.
```bash
git --version
git clone URL_DE_VOTRE_DEPOT cours-python
cd cours-python
git status
```
Ouvrez ensuite le notebook depuis ce dossier. `git status` décrit les changements ; il ne les enregistre pas.

### Au début des séances suivantes
Depuis le dossier déjà cloné, avec votre travail précédent enregistré dans un commit :
```bash
git status
git pull --ff-only
```
`pull` récupère puis intègre les changements du dépôt distant suivi. L'option `--ff-only` demande une mise à jour directe de l'historique ; si les versions ont divergé, Git s'arrête pour que nous examinions la situation.

**Un fork ne reçoit pas automatiquement les nouveautés du dépôt enseignant.** Ce `pull` récupère votre distant habituel, généralement `origin`. Pour les mises à jour du cours, suivez la procédure de synchronisation indiquée par l'enseignant. Si Git signale un conflit ou refuse l'opération, gardez le message et demandez de l'aide avant de modifier les fichiers.


### Après une modification : préparer, enregistrer, envoyer
Sauvegardez d'abord votre notebook dans Jupyter. Dans cet exemple, il est à la racine du dépôt ; adaptez le chemin s'il se trouve dans un sous-dossier.
```bash
git status
git add NB01_Introduction_a_Python.ipynb
git commit -m "NB01 : premiers programmes et manipulations"
git push
git status
```
| Commande | Ce qu'elle fait |
|---|---|
| `clone` | Crée une copie locale d'un dépôt distant. |
| `pull` | Récupère et intègre les changements du distant suivi. |
| `status` | Indique l'état des fichiers et des changements préparés. |
| `add` | Sélectionne la version actuelle d'un fichier pour le prochain commit. |
| `commit` | Enregistre localement les changements préparés. |
| `push` | Envoie les commits au dépôt distant. |

Modifier de nouveau un fichier après `add` demande un nouvel `add` pour inclure cette dernière modification. `push` n'envoie pas les modifications qui ne sont pas dans un commit.

Si Git demande votre identité pour le premier commit, configurez-la **dans ce dépôt**, avec vos informations :

## 6. Les quatre types fondamentaux
Le **type** d'une valeur détermine les opérations qu'on peut lui appliquer. `type(valeur)` permet de l'observer.

| Type | Ce qu'il représente | Exemples |
|---|---|---|
| `int` | Un nombre entier | `6`, `0`, `-12` |
| `float` | Un nombre à virgule flottante, généralement approché | `2.5`, `-0.2`, `3.0` |
| `str` | Du texte, appelé chaîne de caractères | `"rover"`, `"12"`, `""` |
| `bool` | Une valeur logique | `True`, `False` |

Les décimales s'écrivent avec un **point**. `3` et `3.0` ont des types différents. `"3"` est du texte, même s'il contient un chiffre.

Les booléens s'écrivent avec une majuscule et sans guillemets. `"False"` est du texte. Nous utiliserons les booléens pour prendre des décisions dans le prochain notebook ; ici, nous les identifions et les affichons.


````python
nombre_roues = 6
vitesse_m_s = 2.5
nom_rover = "Curieux"
capteur_actif = True
print(nombre_roues, type(nombre_roues))
print(vitesse_m_s, type(vitesse_m_s))
print(nom_rover, type(nom_rover))
print(capteur_actif, type(capteur_actif))
````

### Même symbole, opérations différentes
Avec des nombres, `+` additionne. Avec deux textes, il les assemble : c'est la **concaténation**. `*` peut multiplier des nombres ou répéter un texte un nombre entier de fois.

Prédisez la valeur **et le type** de chaque expression ci-dessous.


````python
print(3 + 4)
print("3" + "4")
print("ha" * 3)
print(type(3 + 4))
print(type("3" + "4"))
````

## 7. Convertir une valeur
Une **conversion** produit une valeur d'un autre type lorsque cela a un sens.

| Écriture | Résultat | Sens |
|---|---|---|
| `int("12")` | `12` | Lire un entier écrit en texte. |
| `float("12.5")` | `12.5` | Lire un nombre décimal écrit en texte. |
| `str(12)` | `"12"` | Produire sa représentation textuelle. |
| `int(12.9)` | `12` | Tronquer la partie décimale vers zéro. |

`int()` **n'arrondit pas** : `int(-12.9)` donne `-12`. `int("12.5")` échoue : le texte ne représente pas directement un entier. Une conversion ne modifie pas la variable source à moins de réaffecter le résultat.


````python
mesure_texte = "12.5"
mesure_m = float(mesure_texte)
print(mesure_texte, type(mesure_texte))
print(mesure_m, type(mesure_m))
print("Mesure : " + str(mesure_m) + " m")
# Avec print, on peut aussi séparer les éléments par des virgules.
print("Mesure :", mesure_m, "m")
````

## 8. Expressions, opérateurs et priorités
Une **expression** combine des valeurs, des noms et des opérateurs pour produire une valeur. Dans `temps_s = distance_m / vitesse_m_s`, l'expression à droite est calculée avant l'affectation.

| Opérateur | Opération | Exemple | Résultat |
|---|---|---|---|
| `+` | Addition | `8 + 3` | `11` |
| `-` | Soustraction | `8 - 3` | `5` |
| `*` | Multiplication | `8 * 3` | `24` |
| `/` | Division | `8 / 3` | environ `2.667` |
| `//` | Quotient arrondi vers le bas | `8 // 3` | `2` |
| `%` | Reste associé à ce quotient | `8 % 3` | `2` |
| `**` | Puissance | `8 ** 3` | `512` |

Avec nos nombres `int` et `float`, `/` produit un `float`, même lorsque la division tombe juste. Avec deux entiers, `//` donne un entier ; avec un opérande `float`, son résultat est un `float`.

Pour des entiers positifs, `//` et `%` décomposent une quantité en groupes complets et reste. Avec `a = 17` et `b = 5` : `17 = 5 × 3 + 2`, donc `a // b` vaut 3 et `a % b` vaut 2. Le diviseur doit être non nul.


````python
print(17 / 5)
print(17 // 5)
print(17 % 5)
print(10 / 2, type(10 / 2))
print(10 // 2, type(10 // 2))
print(10.0 // 2, type(10.0 // 2))
````

### Priorités : rendre le calcul lisible
Pour les expressions usuelles : **parenthèses → puissances → multiplications et divisions (`* / // %`) → additions et soustractions**. Les opérations de même priorité se font de gauche à droite, sauf les puissances qui se regroupent à droite.

La multiplication doit être écrite : `2 * distance_m`, pas `2distance_m`. La puissance s'écrit `**`, pas `^`.

**Prédisez** les quatre résultats puis expliquez l'effet des parenthèses.


````python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(18 / 3 * 2)
print(18 / (3 * 2))
````
